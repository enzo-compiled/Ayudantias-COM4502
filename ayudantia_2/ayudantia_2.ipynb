### Clases, objetos e instancias

- Una **clase** es un molde/plantilla que define atributos y comportamientos.
- Un **objeto** (o **instancia**) es una realización concreta de esa clase, con su propio estado en memoria.
- Se pueden crear tantas instancias como se quiera a partir de una misma clase; cada una es independiente entre sí.

```python
class Perro:
    especie = "Canis lupus familiaris"  #atributo de clase (compartido por todas las instancias)

    def ladrar(self):
        return "¡Guau!"

#Instancias (objetos) de la clase Perro
max = Perro()
toby = Perro()

print(type(max))
print(isinstance(max, Perro))
print(max is toby)
print(max.especie, toby.especie)
```

### El constructor `__init__`

- `__init__` es el método especial que Python llama automáticamente al crear una instancia.
- El primer parámetro siempre es `self`: la referencia al propio objeto. Nótese que puede ser cualquier palabra (no reservada), pero por convención se usa `self`.

```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def saludar(self):
        return f"Hola, soy {self.nombre} y tengo {self.edad} años."

p1 = Persona("Ana", 20)
p2 = Persona("Luis", 25)
print(p1.saludar())
print(p2.saludar())
```

### Métodos especiales

Son métodos con doble guion bajo (`__algo__`) que Python invoca automáticamente en ciertas operaciones. Permiten que los objetos propios se comporten como los tipos nativos (`int`, `str`, `list`, etc.).

Algunos de los más usados:

- `__init__` : constructor/inicializador
- `__str__` / `__repr__` : representación en texto (para el usuario / para el desarrollador)
- `__eq__`, `__lt__`, `__gt__`, ... : comparaciones (`==`, `<`, `>`)
- `__add__`, `__sub__`, `__mul__` : operadores aritméticos (`+`, `-`, `*`)
- `__len__` : soporte para `len(obj)`
- `__getitem__` / `__iter__` / `__next__` : indexado e iteración
- `__call__` : permite usar la instancia como si fuera una función

```python
class Punto:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __repr__(self):
        return f"Punto({self.x}, {self.y})"  #representación código

    def __eq__(self, otro):
        return (self.x, self.y) == (otro.x, otro.y)

    def __add__(self, otro):
        return Punto(self.x + otro.x, self.y + otro.y)

a, b = Punto(1, 2), Punto(3, 4)
print(a + b) # usa __add__  -> Punto(4, 6)
print(a == Punto(1, 2)) # usa __eq__
print(a) # usa __repr__ -> Punto(1, 2)
```

### Herencia y Composición

- **Herencia** modela una relación **"es un/a"** (*is-a*): una subclase extiende/especializa a una clase base con `class Hija(Base):` y `super()`.
- **Composición** modela una relación **"tiene un/a"** (*has-a*): una clase incluye instancias de otras clases como atributos y delega en ellas.

```python
#Herencia
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

    def hacer_sonido(self):
        return "..."

class Perro(Animal):
    def __init__(self, nombre, edad):
      super().__init__(nombre) #herencia de atributos
      self.edad = edad #atributo particular de la clase hija

    def hacer_sonido(self): #sobrescribe el método del padre
        return "Guau"

    def correr(self):
        return "Corriendo"

#Composición:
class Motor:
    def __init__(self, potencia_hp):
        self.potencia_hp = potencia_hp

    def encender(self):
        return f"Motor de {self.potencia_hp}HP encendido"

class Auto:
    def __init__(self, modelo, motor):
        if not modelo:
            raise ValueError("El modelo no puede estar vacio")
        self.modelo = modelo
        self.motor = motor #el Auto delega en su Motor, no hereda de él

    def arrancar(self):
        return f"{self.modelo}: {self.motor.encender()}"

print(Perro("Hueso", 10).hacer_sonido())
print(Auto("Subaru", Motor(150)).arrancar())
```

### Extra: Decoradores

Un decorador es una función que toma otra función, le añade alguna funcionalidad extra, y la devuelve modificada. Es una forma muy limpia de "potenciar" las funciones sin tener que reescribir su código.

```python
#definimos un decorador
def mi_decorador(funcion_original):

    #Esta es la "envoltura" que añade cosas antes y después
    def envoltura():
        print("1. [Decorador] Hago algo ANTES de que la función corra.")

        funcion_original() #Aquí ejecutamos la función real

        print("3. [Decorador] Hago algo DESPUÉS de que la función corra.")

    #El decorador devuelve la envoltura lista para usarse
    return envoltura
```

```python
#Aplicamos el decorador usando el @
@mi_decorador
def saludar():
    print("2. [Función] ¡Hola! ¿Cómo estás?")

saludar()
```

Existen varios decoradores `build-in` (lo dejo a su investigación). Por ejemplo, `@property`, `@total_ordering`, etc.

- Use el decorador `functools.total_ordering` para obtener `<=`, `>` y `>=` "gratis" a partir de solo `__eq__` y `__lt__`. Es decir, solo necesita implementar estos últimos dos.

```python
import math
from functools import total_ordering

@total_ordering
class Fraccion:
    def __init__(self, numerador, denominador):
        pass
```

```python
f1 = Fraccion(1, 2)
f2 = Fraccion(2, 4) # se simplifica a 1/2
f3 = Fraccion(1, 3)

print(f2) # 1/2  (ya simplificada)
print(f1 + f3) # 5/6
print(f1 == f2) # True
print(f3 < f1) # True
print(f1 <= f2, f1 > f3)# True True (gracias a @total_ordering)
```
