## 1. Programación Orientada a Objetos (POO)

* **Clase y objeto:** una clase define qué datos (atributos) y qué comportamiento (métodos) tendrán sus objetos. Un objeto es una instancia concreta de esa clase.
* **Constructor y atributos:** `__init__` inicializa cada instancia. `self` es una referencia al objeto sobre el que se llama el método.
* **Métodos:** consultan o modifican el estado del objeto.
* **Variable local vs atributo:** la local vive solo durante la ejecución del método; el atributo pertenece al objeto y persiste entre llamadas.
* **Atributo de instancia vs de clase:** el de instancia es propio de cada objeto (`self.x`); el de clase se comparte entre todos (`Clase.x`).
* **Métodos especiales útiles:** `__str__` (texto legible), `__repr__` (representación para depurar), `__eq__` (comparación), etc.

### Ejemplo
```
class Cuenta:
    def __init__(self, titular, saldo=0):
        self.titular = titular  
        self.saldo = saldo

    def depositar(self, monto):
        nuevo = self.saldo + monto # variable local: desaparece al terminar
        self.saldo = nuevo # atributo: persiste

    def __str__(self):
        return f"{self.titular}: ${self.saldo}"

c = Cuenta("Juan")
c.depositar(500)
print(c)
```

## 2. Recursividad

* **Definición:** técnica en la que una función resuelve un problema llamándose a sí misma sobre una versión más pequeña del mismo problema.
* **Componentes:**
  * **Caso base:** se resuelve sin nueva llamada recursiva (evita la recursión infinita).
  * **Caso recursivo:** expresa la solución en términos de la misma función.
  * **Progreso:** cada llamada debe acercarse al caso base.
* **Pila de llamadas:** cada llamada ocupa un marco en la pila. En Python el límite por defecto es ~1000 (`sys.getrecursionlimit()`), y **no** optimiza la recursión de cola.
* **Recursividad vs recurrencia:** la recursividad describe *cómo está construido* el algoritmo; la recurrencia describe *matemáticamente su costo*; Big $\mathcal{O}$ expresa cómo crece ese costo con el tamaño de la entrada (cota superior asintótica).

### Ejemplo
```
def factorial(n):
    if n == 0:     #caso base
        return 1
    return n * factorial(n - 1) #caso recursivo
```
Recurrencia: $T(n) = T(n-1) + O(1)  →  T(n) = \mathcal{O}(n)$

### Recurrencias típicas
| Recurrencia            | Complejidad       | Ejemplo               |
|------------------------|----------------|-----------------------|
| T(n) = T(n-1) + O(1)   | O(n)           | factorial             |
| T(n) = T(n/2) + O(1)   | O(log n)       | búsqueda binaria      |
| T(n) = 2T(n/2) + O(n)  | O(n log n)     | merge sort            |
| T(n) = 2T(n-1) + O(1)  | O(2ⁿ)          | Fibonacci ingenuo     |

* **Recursión vs iteración:** toda recursión puede escribirse iterativamente (con pila explícita); la recursiva suele ser más clara, la iterativa más eficiente en memoria.

## 3. Algoritmos de Fuerza Bruta

* **Definición:** resuelve un problema examinando sistemáticamente todos los candidatos relevantes, sin una propiedad que permita descartar la mayoría de antemano.
* **Análisis de costo:** costo total ≈ (número de candidatos) × (costo de evaluar cada uno).
* **Ventajas:** simple, fácil de probar que es correcta, sirve como *baseline* para validar soluciones más eficientes.
* **Desventajas:** el número de candidatos suele crecer exponencialmente o factorialmente.

### Ejemplos de costo
| Problema                     | Candidatos | Costo por candidato | Total        |
|------------------------------|-----------|---------------------|--------------|
| Subset sum (n elementos)     | 2ⁿ        | O(n)                | O(n·2ⁿ)      |
| Permutaciones / TSP          | n!        | O(n)                | O(n·n!)      |
| Buscar patrón en texto       | n − m + 1 | O(m)                | O(n·m)       |
| Contraseña de k dígitos      | 10ᵏ       | O(1)                | O(10ᵏ)       |

### Cómo mejorarla
* **Poda / backtracking:** descartar ramas apenas se sabe que no llevan a solución.
* **Aprovechar estructura:** ordenar, usar dos punteros, hashing o programación dinámica.
* **Cuándo usarla igual:** entradas pequeñas o cuando no se conoce algo mejor.

### Ejemplo
```
def subset_sum(nums, objetivo):
    n = len(nums)
    for mascara in range(1 << n):  # 2^n candidatos
        suma = sum(nums[i] for i in range(n) if mascara >> i & 1)  # O(n)
        if suma == objetivo:
            return True
    return False
```
