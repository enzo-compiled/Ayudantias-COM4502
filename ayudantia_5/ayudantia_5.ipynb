# Método Voraz
Un algoritmo voraz (o *greedy*) es un paradigma de diseño de algoritmos que construye una solución paso a paso. Consiste en resolver un problema mediante una secuencia de soluciones localmente óptimas. Consideramos que un algoritmo es voraz si en cada paso realizado la solución local es:
- Factible (satisface las restricciones del problema).
- Localmente óptima.
- Irrevocable.

## Ventajas y Desventajas
* **Ventajas:** Suelen ser intuitivos y fáciles de implementar y generalmente muy eficientes en cuanto a tiempo de ejecución (suelen estar dominados por la complejidad de ordenar los datos).
* **Desventajas:** No siempre garantizan la solución óptima global.

## Ejercicios

1. **Selección de Actividades:**
   * Tienes $n$ actividades, cada una con un tiempo de inicio $s_i$ y un tiempo de fin $f_i$. ¿Cuál es el número máximo de actividades no superpuestas que una persona puede realizar? Es decir, $s_{i+1} \gt f_i$.

```python
#considere cada par [inicio, fin] como una actividad.
inicio_act = [1, 3, 0, 5, 8, 5]
fin_act = [2, 4, 6, 7, 9, 9]

def seleccion_actividades(inicio, fin):
    #ordenar por fin
    actividades = sorted(zip(inicio, fin), key=lambda a: a[1])
    seleccionadas = []
    ultimo_fin = -1

    for s, f in actividades:
        if s > ultimo_fin:
            seleccionadas.append((s, f))
            ultimo_fin = f

    return len(seleccionadas), seleccionadas

cantidad, acts = seleccion_actividades(inicio_act, fin_act)
print(cantidad)
print(acts)
```

2. **Supervivencia en una isla**:
* Juan quedó atrapado en una isla. Solo hay una tienda en esta isla y está abierta todos los días de la semana excepto los domingos. Considera las siguientes restricciones:

*   **N** – La cantidad máxima de unidades de comida que puedes comprar cada día.
*   **S** – Número de días que necesitas sobrevivir.
*   **M** – Unidades de comida requeridas cada día para sobrevivir.

Actualmente es lunes y necesita sobrevivir durante los próximos S días; inicialmente no tiene comida.

Encuentra el número mínimo de días en los que se necesita comprar comida en la tienda para que pueda sobrevivir los próximos S días. Si no es posible sobrevivir durante los S días, entonces devuelve -1.

Ejemplo:
- Input: S = 10, N = 16, M = 2
- Output: 2


```python
#completar
def Sobrevivir(S, N, M):
    comida_necesaria = S * M
    domingos = S // 7
    dias_compra_posibles = S - domingos

    if M > N or dias_compra_posibles * N < comida_necesaria:
        return -1

    return (comida_necesaria + N - 1) // N


print(Sobrevivir(10, 16, 2))
print(Sobrevivir(10, 1, 2))
```

# Introducción a Grafos
Un **grafo** es una estructura matemática y computacional que permite modelar elementos y las relaciones entre ellos. Suele denotarse como $G=(V,E)$ y consta de dos componentes fundamentales:
* **Vértices (o nodos):** Representan los elementos individuales del conjunto ($V$).
* **Aristas (o conexiones):** Representan los enlaces o relaciones entre los vértices ($E$).

Dependiendo de sus características, un grafo puede ser **dirigido** (si las conexiones tienen un sentido definido) o **no dirigido** (si la relación funciona en ambos sentidos), y puede ser **ponderado** (si cada arista tiene un valor o costo) o **no ponderado**.

## Conceptos

- **Adyacencia:** Dos vértices son adyacentes si están conectados directamente por una arista.
- **Grado:** Es la cantidad de aristas incidentes en un vértice específico (en un grafo no dirigido).
- **Camino:** Es una secuencia de vértices que están conectados por aristas.
- **Ciclo:** Es un camino que regresa al punto inicial sin repetir otros vértices en el trayecto.
- **Árbol:** Es un caso especial de grafo que es no dirigido, conexo (hay camino entre cada par de vértices) y no tiene ciclos.
